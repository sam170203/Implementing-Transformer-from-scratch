```python
import torch
import torch.nn.functional as F
import math

def scaled_dot_product_attention(Q: torch.Tensor, K: torch.Tensor, V: torch.Tensor) -> torch.Tensor:
    """
    Compute scaled dot-product attention.
    """
    # Yo lesss do it
    d_k=Q.size(-1)  #dimensions of key vectors

    #now QK^transpose
    scores=torch.matmul(Q,K.transpose(-2,-1))
    scores=scores/math.sqrt(d_k)

    #now softmax -> attention weights

    attn_weights=F.softmax(scores,dim=-1)
    output=torch.matmul(attn_weights,V)
    return output
```

```python
torch.set_printoptions(precision=3, sci_mode=False)

Q = torch.tensor([[[1., 0., 1.],
                   [0., 1., 1.]]])

K = torch.tensor([[[1., 1., 0.],
                   [0., 1., 1.],
                   [1., 0., 1.]]])

V = torch.tensor([[[1., 0.],
                   [0., 2.],
                   [3., 1.]]])

d_k = Q.size(-1)

scores = torch.matmul(Q, K.transpose(-2, -1))
scaled = scores / math.sqrt(d_k)
weights = F.softmax(scaled, dim=-1)
out = torch.matmul(weights, V)

print("Q =\n", Q)
print("\nK =\n", K)
print("\nV =\n", V)

print("\nQK^T =\n", scores)
print("\nScaled scores =\n", scaled)
print("\nAttention weights =\n", weights)
print("\nOutput =\n", out)
```
