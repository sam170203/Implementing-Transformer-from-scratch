```python
# ==========================================
# Multi Head Attention Implementation
# From Scratch using NumPy
# ==========================================

import numpy as np

np.random.seed(42)


# Softmax Function


def softmax(x, axis=-1):
    """
    Numerically stable softmax
    """
    e_x = np.exp(x - np.max(x, axis=axis, keepdims=True))
    return e_x / np.sum(e_x, axis=axis, keepdims=True)



# Multi Head Attention Implementation


def multi_head_attention(Q, K, V, W_q, W_k, W_v, W_o, num_heads):
    """
    Multi Head Attention from scratch.

    Shapes:
    Q,K,V : (batch_size, seq_len, d_model)
    W_q,W_k,W_v : (d_model, d_model)
    W_o : (d_model, d_model)
    """

    batch_size, seq_len, d_model = Q.shape

    assert d_model % num_heads == 0, "d_model must be divisible by num_heads"

    d_k = d_model // num_heads


    # Step 1: Linear projections


    Q = Q @ W_q
    K = K @ W_k
    V = V @ W_v


    # Step 2: Split heads


    Q = Q.reshape(batch_size, seq_len, num_heads, d_k).transpose(0,2,1,3)
    K = K.reshape(batch_size, seq_len, num_heads, d_k).transpose(0,2,1,3)
    V = V.reshape(batch_size, seq_len, num_heads, d_k).transpose(0,2,1,3)


    # Step 3: Scaled Dot Product Attention


    scores = Q @ K.transpose(0,1,3,2)

    scores = scores / np.sqrt(d_k)

    attention_weights = softmax(scores)

    attention_output = attention_weights @ V


    # Step 4: Concatenate heads


    attention_output = attention_output.transpose(0,2,1,3)

    attention_output = attention_output.reshape(batch_size, seq_len, d_model)


    # Step 5: Final projection


    output = attention_output @ W_o

    return output



# Example Run


batch_size = 2
seq_len = 4
d_model = 8
num_heads = 2

# Random inputs (simulating token embeddings)

Q = np.random.randn(batch_size, seq_len, d_model)
K = np.random.randn(batch_size, seq_len, d_model)
V = np.random.randn(batch_size, seq_len, d_model)

# Weight matrices

W_q = np.random.randn(d_model, d_model)
W_k = np.random.randn(d_model, d_model)
W_v = np.random.randn(d_model, d_model)
W_o = np.random.randn(d_model, d_model)

# Run Multi Head Attention

output = multi_head_attention(Q, K, V, W_q, W_k, W_v, W_o, num_heads)

print("Input Shape:", Q.shape)
print("Output Shape:", output.shape)

print("\nOutput Tensor:\n")
print(output)
```
